```python
# negative values
numeric_columns = df.select_dtypes(
    include=np.number
).columns

negative_values = {}

for column in numeric_columns:
    count = (df[column] < 0).sum()
    if count > 0:
        negative_values[column] = count

negative_values
```

```python
# zero values
zero_values = {}

for column in numeric_columns:
    count = (df[column] == 0).sum()
    if count > 0:
        zero_values[column] = count

zero_values
```

```python
correlation_columns = [
    "Soil_pH",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Organic_Carbon_pct",
    "Soil_Moisture_pct",
    "Avg_Temperature_C",
    "Rainfall_mm",
    "Humidity_pct",
    "Farm_Area_Acres",
    "Fertilizer_kg_acre",
    "Yield_kg_acre",
    "Production_kg",
    "Total_Cost_Rs",
    "Revenue_Rs",
    "Profit_Rs"
]

correlation_matrix = df[correlation_columns].corr()

correlation_matrix
```

```python
# Correlation heatmap
plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Agricultural Dataset Correlation Matrix")
plt.tight_layout()

plt.show()
```

```python
outlier_summary = []

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": len(outliers) / len(df) * 100
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df.sort_values(
    by="Outlier_Count",
    ascending=False
)
```

```python
# Boxplots for important variables
important_numeric_columns = [
    "Soil_pH",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Rainfall_mm",
    "Yield_kg_acre",
    "Profit_Rs"
]

for column in important_numeric_columns:

    plt.figure(figsize=(10, 5))

    sns.boxplot(x=df[column])

    plt.title(f"Outlier Analysis - {column}")
    plt.xlabel(column)

    plt.tight_layout()
    plt.show()
```

```python
df["Calculated_Profit"] = (
    df["Revenue_Rs"] - df["Total_Cost_Rs"]
)

df["Profit_Difference"] = (
    df["Profit_Rs"] - df["Calculated_Profit"]
)

df["Profit_Difference"].describe()
```

```python
df["Calculated_Production"] = (
    df["Yield_kg_acre"] *
    df["Farm_Area_Acres"]
)

df["Production_Difference"] = (
    df["Production_kg"] -
    df["Calculated_Production"]
)

df["Production_Difference"].describe()
```
